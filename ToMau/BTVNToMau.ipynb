```python
import random

classes = ["10A1", "10A2"]

subjects = {
    "Toán": "GV1",
    "Tin": "GV2"
}

days = ["T2", "T3", "T4", "T5", "T6"]

periods = [1, 2, 3, 4, 5]


required_lessons = {
    "Toán": 5,
    "Tin": 3
}

def create_individual():

    timetable = []

    for cls in classes:

        for subject, total in required_lessons.items():

            count = 0

            while count < total:

                day = random.choice(days)

                period = random.choice(periods)

                gene = (cls, day, period, subject)

                if gene not in timetable:

                    timetable.append(gene)

                    count += 1

    return timetable

def fitness(individual):

    penalty = 0

    used = {}

    for gene in individual:

        cls, day, period, subject = gene

        teacher = subjects[subject]

        key_class = (cls, day, period)

        key_teacher = (teacher, day, period)

        if key_class in used:
            penalty += 10
        else:
            used[key_class] = 1

        if key_teacher in used:
            penalty += 10
        else:
            used[key_teacher] = 1

    return 1 / (1 + penalty)


def crossover(parent1, parent2):

    point = random.randint(1, len(parent1)-1)

    child = parent1[:point] + parent2[point:]

    return child

def mutate(individual):

    idx = random.randint(0, len(individual)-1)

    cls, day, period, subject = individual[idx]

    new_day = random.choice(days)

    new_period = random.choice(periods)

    individual[idx] = (cls, new_day, new_period, subject)

    return individual

population_size = 50

population = []

for i in range(population_size):

    population.append(create_individual())

generations = 100

for gen in range(generations):

    population = sorted(
        population,
        key=lambda x: fitness(x),
        reverse=True
    )

    new_population = population[:10]

    while len(new_population) < population_size:

        parent1 = random.choice(population[:20])

        parent2 = random.choice(population[:20])

        child = crossover(parent1, parent2)

        if random.random() < 0.2:

            child = mutate(child)

        new_population.append(child)

    population = new_population


best = population[0]

print("THỜI KHÓA BIỂU TỐT NHẤT:\n")

for item in best:

    cls, day, period, subject = item

    teacher = subjects[subject]

    print(
        "Lớp:",
        cls,
        "|",
        day,
        "| Tiết",
        period,
        "|",
        subject,
        "| Giáo viên:",
        teacher
    )

print("\nFitness =", fitness(best))
```
