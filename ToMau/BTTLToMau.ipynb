```python

# BÀI 2
# Chia chương trình thành các hàm

from google.colab import files


def read_graph():

    uploaded = files.upload()

    file_name = list(uploaded.keys())[0]

    G = []

    with open(file_name, 'r') as f:

        for line in f:
            row = list(map(int, line.strip().split()))
            G.append(row)

    return G


def create_nodes(n):

    node = ""

    for i in range(n):
        node += chr(65 + i)

    return node


def calculate_degree(G):

    degree = []

    for i in range(len(G)):
        degree.append(sum(G[i]))

    return degree


def sort_nodes(node, degree):

    sortedNode = []
    indeks = []

    for i in range(len(degree)):

        _max = -1

        for j in range(len(degree)):

            if j not in indeks:

                if degree[j] > _max:
                    _max = degree[j]
                    idx = j

        indeks.append(idx)
        sortedNode.append(node[idx])

    return sortedNode


def graph_coloring(G, node, sortedNode):

    colors = ["Blue", "Red", "Yellow", "Green", "Orange"]

    t_ = {}

    for i in range(len(node)):
        t_[node[i]] = i

    colorDict = {}

    for i in range(len(node)):
        colorDict[node[i]] = colors.copy()

    solution = {}

    for nd in sortedNode:

        availableColors = colorDict[nd]

        solution[nd] = availableColors[0]

        adjacentNode = G[t_[nd]]

        for j in range(len(adjacentNode)):

            if adjacentNode[j] == 1:

                if availableColors[0] in colorDict[node[j]]:
                    colorDict[node[j]].remove(availableColors[0])

    return solution

def print_solution(solution):

    print("\nKẾT QUẢ TÔ MÀU:\n")

    for vertex, color in sorted(solution.items()):
        print("Đỉnh", vertex, "=", color)


G = read_graph()

n = len(G)

node = create_nodes(n)

degree = calculate_degree(G)

sortedNode = sort_nodes(node, degree)

solution = graph_coloring(G, node, sortedNode)

print_solution(solution)
```

```python

# BÀI 3
# Thuật toán người bán hàng (TSP)

from itertools import permutations


G = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

cities = ["A", "B", "C", "D"]

start = 0

vertex = []

for i in range(len(G)):
    if i != start:
        vertex.append(i)

min_path = float('inf')

best_path = []

for path in permutations(vertex):

    current_cost = 0

    k = start
    for j in path:

        current_cost += G[k][j]
        k = j
    current_cost += G[k][start]
    if current_cost < min_path:

        min_path = current_cost

        best_path = [start] + list(path) + [start]

print("ĐƯỜNG ĐI TỐI ƯU:")

for i in best_path:
    print(cities[i], end=" ")

print("\n")

print("CHI PHÍ NHỎ NHẤT =", min_path)
```
